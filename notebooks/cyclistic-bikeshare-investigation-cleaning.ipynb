Removing the duplicates

```python
bike_df = bike_df.drop_duplicates(
    subset=['ride_id']
)
```

```python
bike_df["ride_id"].duplicated().sum()
```

Change format from text to datetime

```python
# convert started_at and ended_at to datetime format
bike_df["started_at"] = pd.to_datetime(bike_df["started_at"])
bike_df["ended_at"] = pd.to_datetime(bike_df["ended_at"])
```

```python
# verify data types
bike_df[["started_at","ended_at"]].dtypes
```

Added new column ride_length

```python
# add new column 'ride_length' to calculate difference between ended_at and started_at
bike_df["ride_length"] = (
    bike_df["ended_at"] -
    bike_df["started_at"]
).dt.total_seconds() / 60
```

```python
bike_df["ride_length"].describe()
```

Investigating negative, zero, rides more than a day

```python
# negative rides
len(bike_df[bike_df["ride_length"] < 0])
```

```python
# zero rides
len(bike_df[bike_df["ride_length"] == 0])
```

```python
#very long rides (24+ hours)
bike_df[bike_df["ride_length"] > 1440]
```

Create date features: date, month, day of the week, hour

```python
# Date
bike_df["date"] = bike_df["started_at"].dt.date
```

```python
# month
bike_df["month"] = bike_df["started_at"].dt.month_name()
```

```python
# day of the week
bike_df["day_of_week"] = bike_df["started_at"].dt.day_name()
```

```python
# hour
bike_df["hour"] = bike_df["started_at"].dt.hour
```

Reinvestigate possible ride_length issues

```python
# Remove negative ride lengths (29)
bike_df = bike_df[bike_df["ride_length"] >= 0]
```

Ride over 24+ hours investigation

```python
long_rides = bike_df[bike_df["ride_length"] > 1440]
long_rides["rideable_type"].value_counts()
```

```python
long_rides["member_casual"].value_counts()
```

```python
long_rides.groupby("rideable_type")["ride_length"].describe()
```

Proceed removing > 24 hrs ride_length

```python
# only keep ride_length records that are not negative and does not exceed 24 hours
bike_df = bike_df[
    (bike_df["ride_length"] >= 0)
    &
    (bike_df["ride_length"] <= 1440)
]
```

```python
# keep missing station name and id, remove missing lat and lng
bike_df = bike_df.dropna(
    subset=["end_lat", "end_lng"]
)
```
